```python
class Solution:
    def firstSearch(self, arr, k):
        # Code Here
        left = 0
        right = len(arr) - 1
        result = -1
        
        while left <= right:
            mid = left + (right - left) // 2
            
            if arr[mid] == k:
                result = mid
                right = mid - 1
            elif arr[mid] < k:
                left = mid + 1
            else:
                right = mid - 1
        
        return result

cases = [
    ([1,2,3,4,5], 4),
    ([11,22,33,44,55],445),
    ([1,1,1,1,2],1)
]

for arr, k in cases:
    res = Solution().firstSearch(arr, k)
    print(f"Input: arr[] = {arr}, k = {k}\n Output: result = {res}")
```
